### Regras Gramaticais Aprendidas

#### Preposições: TO vs. FOR
- **TO** → objetivo ou direção (seta)
  - I want to improve my English.
  - Send this to João.
- **FOR** → destinatário ou propósito (bandeja)
  - This is for you.
  - I study for my exam.

#### Preposições de Local: IN, ON, AT
- **ON** → superfície
  - The book is on the table.
- **IN** → ambiente ou grupo
  - I work in a tech company.
- **AT** → local físico específico
  - Meet me at the café.

#### Ordem dos Adjetivos
Características ANTES do substantivo.
- black car (não "car black")
- speaking lessons (não "lessons speaking")

#### Pronomes: HE/SHE vs. IT
- **HE/SHE** → pessoas e pets
  - João is smart. He works as a developer.
  - My dog is friendly. She loves to play.
- **IT** → tudo o mais (objetos, conceitos)
  - The project is complex. It requires planning.

#### Regra de Sufixo: -ÇÃO → -TION
- Gesticulação → Gesticulation
- Comunicação → Communication
- Apresentação → Presentation

#### Infinitivos (TO + VERBO)
Ação sem sujeito ou tempo específico.
- I want to improve.
- It's time to leave.

### Vocabulário

- **Hiring Processes** (processo de contratação) — The international hiring processes are complex.

## Referência de Gramática

### Preposições Principais
- **TO** → objetivo/direção
- **FOR** → destinatário/propósito
- **IN** → ambiente/grupo
- **ON** → superfície
- **AT** → local específico

### Pronomes
- **He/She** → pessoas e pets
- **It** → objetos, conceitos, eventos

### Ordem de Palavras
- Adjetivos VÊM ANTES do substantivo
- Subject → Verb → Object (SVO)

### Sufixos Comuns
- -ção → -tion (português → inglês)
